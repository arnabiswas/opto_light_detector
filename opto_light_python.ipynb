```python
lower_blue = np.array([110,50,50])
upper_blue = np.array([130,255,255])

start_cropx, end_cropx = 50, 1100

def detect_blue_light(img):
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    mask = cv2.inRange(hsv, lower_blue, upper_blue)
    res = cv2.bitwise_and(img, img, mask=mask)
    contours, _ = cv2.findContours(mask, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

    # If there are any contours that means we have detected blue color
    if len(contours)>5:
        print("Blue color detected!")
        on_off = 1
    else:
        print("Blue color NOT detected!")
        on_off = 0
        
    return on_off
```

```python
use_frame = frame.copy()
use_frame = use_frame[:,start_cropx:end_cropx,:]
```

```python
text = "On"
text_position = (800, 50)  # (x, y) coordinates, start from top left
font = cv2.FONT_HERSHEY_SIMPLEX
font_scale = 1
color = (255, 0, 0)  # Blue color
thickness = 2

fourcc = cv2.VideoWriter_fourcc(*'XVID') 
out = cv2.VideoWriter('output.avi', fourcc, 30, (use_frame.shape[1], use_frame.shape[0])) # 20.0 is the FPS (frames per second), (640, 480) is the frame size
cap = cv2.VideoCapture('RC6-11-Live.mp4')

on_off_framewise = []
read_sucess = True
while read_sucess:
    read_sucess, frame = cap.read()
    if not read_sucess:
        break
    #crop the frame
    use_frame = frame.copy()
    use_frame = use_frame[:,start_cropx:end_cropx,:]
    on_off = detect_blue_light(use_frame)
    on_off_framewise.append(on_off)
    
    if on_off:
        #write to frame
        use_frame = cv2.putText(use_frame, text, text_position, font, font_scale, color, thickness, cv2.LINE_AA)
    out.write(use_frame)

out.release()
cap.release()
```
